# CodeMonkey RAGAS report
Run `python -m evaluation.ragas_eval --profile deterministic` or the full profile before opening this notebook. Scores shown here come only from the generated report.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

report_path = Path('../python-backend/evaluation/results/ragas_report.json')
report = json.loads(report_path.read_text(encoding='utf-8'))
scores = pd.DataFrame([{'id': row['id'], **row['scores']} for row in report['cases']]).set_index('id')
scores
```

```python
ax = scores.mean(numeric_only=True).sort_values().plot(kind='barh', color='#14b8a6')
ax.set_xlim(0, 1)
ax.set_xlabel('Mean score (higher is better)')
ax.set_title(f"RAGAS metrics — {report['profile']} profile, n={report['case_count']}")
plt.tight_layout()
```
